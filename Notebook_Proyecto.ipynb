# «La Carrera de la Descarbonización: 2000–2025»
### (Enfoque Temporal y Velocidad)

## 📖 Concepto Narrativo
Una competencia sobre qué tan rápido han avanzado (o retrocedido) los países de la América en sustituir las fuentes fósiles por renovables directas.

## 🎯 Pregunta de Diseño
> **¿Quiénes lideran la transición energética en el continente y qué países se quedaron estancados en la era del carbón y el gas?**

**Dataset:** `release_generation_yearly_global.csv` (fuente: Ember Global Electricity Review)

```python
import pandas as pd

# Carga de datos globales de Ember
df_total = pd.read_csv('release_generation_yearly_global.csv')

# Filtración por continentes de las Américas
Cont_America = ['South America', 'North America']
df_america = df_total[df_total['Continent'].isin(Cont_America)]

# Países principales seleccionados (19 países de las Américas)
paises_principales = [
    'Argentina', 'Bolivia', 'Brazil', 'Canada', 'Chile', 'Colombia',
    'Costa Rica', 'Cuba', 'Ecuador', 'Guatemala', 'Honduras', 'Mexico',
    'Nicaragua', 'Panama', 'Paraguay', 'Peru', 'United States', 'Uruguay', 'Venezuela'
]

df_america = df_america[df_america['Area'].isin(paises_principales)]

# Filtración temporal: Período de la Carrera de la Descarbonización (2000 en adelante)
df_america = df_america[df_america['Year'].between(2000, 2025)]

print(f"Dimensiones del dataset filtrado: {df_america.shape}")
print(f"Países presentes ({len(df_america['Area'].unique())}): {sorted(df_america['Area'].unique())}")
print(f"Rango de años: {df_america['Year'].min()} - {df_america['Year'].max()}")
```

## ⚡ Fuentes de Energía y Transición Energética
Filtramos las fuentes directas de energía eléctrica para analizar la evolución anual (generación en TWh y porcentaje de participación).

- **Fuentes Renovables / Limpias:** Solar, Wind, Hydro, Bioenergy, Other renewables (y Nuclear si aplica a limpias)
- **Fuentes Fósiles:** Coal, Gas, Other fossil

```python
# Fuentes individuales (no agregadas)
df_sources = df_america[df_america['Is aggregated source'] == False]

# Resumen anual por país y fuente
df_country_year_source = df_sources[[
    'Area', 'Year', 'Electricity source', 'Generation (TWh)', 'Share of generation (%)'
]].sort_values(by=['Area', 'Year', 'Electricity source'])

# Agregados macro: Renovables vs Fósiles
df_macro = df_america[df_america['Electricity source'].isin(['Renewables', 'Fossil', 'Total generation'])]
df_pivot_share = df_macro.pivot_table(
    index=['Area', 'Year'],
    columns='Electricity source',
    values='Share of generation (%)'
).reset_index()

df_pivot_share
```

## 🔊 Emisiones de CO2 y Datos para Sonificación
Extraemos las emisiones producidas (en MtCO2e) por fuente y las emisiones totales por país para alimentar la narrativa sonora y el diseño de la sonificación:
- A mayor volumen de emisiones fósiles: mayor intensidad o distorsión acústica.
- A mayor sustitución por fuentes renovables: transición hacia timbres armónicos limpios.


```python
# Resumen anual de emisiones por país y fuente (MtCO2e)
df_emissions = df_sources[[
    'Area', 'Year', 'Electricity source', 'Generation (TWh)', 'Emissions (MtCO2e)'
]].dropna().sort_values(by=['Area', 'Year'])

# Emisiones totales anuales por país para sonificación
df_total_emissions = df_america[df_america['Electricity source'] == 'Total generation'][[
    'Area', 'Year', 'Generation (TWh)', 'Emissions (MtCO2e)', 'Emissions intensity (gCO2e/kWh)'
]].sort_values(by=['Area', 'Year'])

df_total_emissions.head(15)
```
